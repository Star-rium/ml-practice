```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Bài 1: Xây dựng bootstrap method using numpy 

```python
data = [58, 70, 82, 95, 63, 77, 88, 69, 73, 85]
n_samples = 2000

conf = 95
rng = np.random.default_rng(seed=63)
bootstraps = np.array([rng.choice(data, size=6, replace=True) for _ in range(n_samples)])

# print(bootstraps[:5])
means = np.mean(bootstraps, axis=1)
overall_mean = np.mean(means)
# print(means[:5])

low_percentile = (100 - 95) / 2
high_percentile = 100 - low_percentile
lower_bound = np.percentile(means, low_percentile)
higher_bound = np.percentile(means, high_percentile)

print(f"{overall_mean}, {lower_bound}, {higher_bound}")

ax = sns.histplot(means)
plt.axvline(overall_mean, color="red")
plt.axvline(lower_bound, color="blue")
plt.axvline(higher_bound, color="blue")
plt.xlabel("Mean values")
plt.ylabel("Counts")
plt.show()
```
